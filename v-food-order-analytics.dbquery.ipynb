```sql
CREATE OR REPLACE VIEW v_food_orders_analytics AS
SELECT 
    Order_ID,
    User_ID,
    Restaurant_ID,
    Driver_ID,
    Item_Name,
    
    -- Numerical Fields
    CAST(Quantity AS INT) AS Quantity,
    CAST(Total_Price AS DECIMAL(10,2)) AS Total_Price,
    CAST(Delivery_Distance_km AS DOUBLE) AS Delivery_Distance_km,
    
    -- Delivery duration handling
    -- Sets duration to NULL for non-delivered orders to avoid skewing average duration metrics
    CASE 
        WHEN Order_Status = 'Delivered' THEN CAST(Delivery_Duration_Minutes AS INT)
        ELSE NULL 
    END AS Delivery_Duration_Minutes,
    
    -- Date transformations
    to_timestamp(Order_Time, 'dd-MM-yyyy HH:mm') AS Order_Time,
    CASE 
        WHEN Order_Status = 'Delivered' THEN to_timestamp(Delivery_Time, 'dd-MM-yyyy HH:mm')
        ELSE NULL 
    END AS Delivery_Time,
    
    -- Categorical Fields
    City,
    Payment_Method,
    Order_Status,
    Driver_Vehicle,
    Traffic_Level,
    Driver_Availability,

    -- Map and coordiantes for mapping 
    CAST(Restaurant_Lat AS DOUBLE) AS Restaurant_Lat,
    CAST(Restaurant_Lon AS DOUBLE) AS Restaurant_Lon,
    CAST(Customer_Lat AS DOUBLE) AS Customer_Lat,
    CAST(Customer_Lon AS DOUBLE) AS Customer_Lon
FROM workspace.food_order_delivery.data;
```
